A container is a data structure that holds multiple elements allowing easy data storage and manipulation.

Some examples of containers are: list, tuple, set, dictionary, etc.

# 🔎 Finding An Item

Any item can be found easily in the container:


```python
item in container
```

## 🧾 List

````python
number_list = [1, 2, 3, 4]
````

````python
4 in number_list
````

````python
5 in number_list
````

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ <code>4</code> is in the <code style="color:green;">number_list</code> but not <code>5</code>. ⤴️
</div>

## 🏷️ Dictionary

````python
number_dictionary = {
    1: "one",
    2: "two",
    3: "three"
}
````

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Find an item in the <code>keys</code> of a dictionary. ⤵️
</div>

````python
1 in number_dictionary
````

````python
4 in number_dictionary
````

````python
"one" in number_dictionary
````

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ <code>1</code> is in the <code style="color:green;">number_dictionary</code> but not <code>4</code> and <code>"one"</code>. ⤴️
</div>

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Find an item in the <code>values</code> of a dictionary. ⤵️
</div>

````python
"one" in number_dictionary.values()
````

````python
"four" in number_dictionary.values()
````

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ <code>"one"</code> is in the <code style="color:green;">number_dictionary</code> but not <code>"four"</code>. ⤴️
</div>

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Another way to find an item in the keys of a dictionary: ⤵️
</div>

````python
1 in number_dictionary.keys()
````

````python
4 in number_dictionary.keys()
````

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ <code>1</code> is in the <code style="color:green;">number_dictionary</code> but not <code>4</code>. ⤴️
</div>

### 📚 Container and Loop

We can create a loop to iterate through the container and retrieve each item.

````python
number_list = [1, 2, 3, 4]

for item in number_list:
    print(item)
````
