An iterator is an object that allows you to traverse through elements of a collection (like lists, tuples, dictionaries, or custom objects) one at a time without needing to know the underlying structure.

* An iterator is a special object that `determines its next state based solely on its current state`.

* To `create an iterator`:
  ```python
  iter()
  ```
* To `determine the next state`:
  ```python
  next()
  ```

### Create custom iterator
A customized iterator can be built if we so desire. To achieve this, we must construct a class and add the `__iter__()` and `__next__()` methods to it. For instance, let's build the class `my_range` which is identical to `range(n)`.

````python
class my_range:
    def __init__(self, index):
        self.index = 0
        self.max_index = index - 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.index <= self.max_index:
            self.index += 1
            return self.index - 1
        else:
            raise StopIteration()


if __name__ == "__main__":
    for i in my_range(5):
        print(i)
````

Note: The `__iter__()` method only returns `self`. However, Python can recognize that this class is an iterator for this method.

If the `for` loop is unable to perform the `iteration` for us, then the following code must be written ⤵️

````python
if __name__ == "__main__":
    it = my_range(5)
    while True:
        try:
            print(next(it))
        except StopIteration:
            break
````

Note: Instead of writing `next()`, we can also write `it. next__()`.

## Iterable
Iterable refers to a function that can produce an iterator. [Strings](), [lists](), [tuples](), [sets](), and [dictionaries]() are all `iterable` in Python.

#### List - Iterable

````python
li = [1, 2, 3]
it = iter(li)
type(it), it
````

#### Tuple - Iterable

````python
tpl = (1, 2, 3)
it = iter(tpl)
type(it), it
````

#### Set - Iterable

````python
st = {1, 2, 3}
it = iter(st)
type(it), it
````

#### Dictionary - Iterable

````python
dt = {"a": "A", "b": "B"}
it = iter(dt)
type(it), it
````
