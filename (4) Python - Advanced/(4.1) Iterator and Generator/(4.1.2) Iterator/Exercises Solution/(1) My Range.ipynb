```python
class my_range:
    def __init__(self, start, stop = None, step = 1):
        self.index = 0 if stop is None else start
        self.step = step
        self.max_index = (start - 1) if stop is None else (stop - 1)

    def __iter__(self):
        return self

    def __next__(self):
        if self.index <= self.max_index:
            self.index += self.step
            return self.index - self.step
        else:
            raise StopIteration()
```

```python
if __name__ == "__main__":
    for i in my_range(10):
        print(i)
```

```python
if __name__ == "__main__":
    for i in my_range(4, 11):
        print(i)
```

```python
if __name__ == "__main__":
    for i in my_range(4, 11, 2):
        print(i)
```
