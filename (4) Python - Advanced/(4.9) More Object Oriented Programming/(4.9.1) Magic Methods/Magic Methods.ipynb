Adding `Car` object

```python
class Car:
    def __init__(self, name, manufacturer, year, price):
        self.name = name
        self.manufacturer = manufacturer
        self.year = year
        self.price = price

    def __add__(self, other):
        return self.price + other.price


if __name__ == "__main__":
    car1 = Car("Camry", "Toyota", 2020, 32000)
    car2 = Car("Camry", "Toyota", 2022, 38000)
    print(car1 + car2)
```

`Car` class

```python
class Car:
    def __init__(self, name, manufacturer, year):
        self.name = name
        self.manufacturer = manufacturer
        self.year = year

    def __repr__(self):
        return "Car ('{}', '{}', '{}')".format(self.name, self.manufacturer, self.year)

    def __str__(self):
        return "{} {} {}".format(self.manufacturer, self.name, self.year)

if __name__ == "__main__":
    car = Car("Camry", "Toyota", 2020)
    print(car)
    print(repr(car))
```

# Property Decorators

```python
class Car:
    def __init__(self, name, manufacturer, year, price = None):
        self.name = name
        self.manufacturer = manufacturer
        self.year = year
        self._price = price

    @property
    def price(self):
        return f"{self._price:.2f}"

    @price.setter
    def price(self, new_price):
        if not isinstance(new_price, (float, int)):
            print("Invalid price")
        elif new_price < 0:
            print("Price cannot be negative")
        else:
            self._price = new_price

    @price.deleter
    def price(self):
        self._price = None


if __name__ == "__main__":
    car = Car("Camry", "Toyota", 2020, 32000.0)
    print(car.price)
    car.price = 35000.123456
    print(car.price)
    car.price = "Invalid price"
    car.price = -10000.78910
    print(car.price)
```
