```python
def find_index(li, n):
    if n <= li[0]:
        return 0
    if n > li[-1]:
        return len(li)

    for i in range(len(li)-1):
        if li[i] < n and li[i+1] >= n:
            return i+1

    # the code should never reach here
    raise("couldn't find the right place for n")
```

```python
def insort(li, n):
    # find the right place (index) for n in li
    i = find_index(li, n)

    # add n at the end of the list
    li.append(n)

    # now shift all items one place right, starting from index i
    for j in range(len(li)-1, i, -1):
        li[j] = li[j-1]

    # now put n into it's correct position
    li[i] = n

    print(li)
```

```python
if __name__ == "__main__":
    for n in range(11):
        li = [1, 2, 2, 3, 4, 6, 8, 9]
        print(li, n)
        insort(li, n)
        print()
```

Exercise: find_index -> use binary search, insort -> use insert() method

```python
# Binary refactoring done for finding indexes
def binary_find_index(items, number):
    if n <= items[0]:
        return 0
    if n > items[-1]:
        return len(items)

    left, right = 0, len(items) - 1
    mid = 0

    while left <= right:
        mid = (left + right) // 2

        if items[mid] == number:
            return mid
        if items[mid] < number:
            left = mid + 1
        else:
            right = mid - 1

    return left
```

```python
# insort function using insert() method
def insert_insort(items, number):
    i = binary_find_index(items, number)
    items.insert(i, number)
    print("List after insertion: ", items)
```

```python
if __name__ == "__main__":
    for n in range(11):
        li = [1, 2, 2, 3, 4, 6, 8, 9]
        print("List: ", li)
        print("Number: ", n)
        insert_insort(li, n)
        print()
```

```python
# insert a number in the sorted list using bisect()
from bisect import bisect


def insort(li, n):
    # find the right place (index) for n in li
    i = bisect(li, n)

    # insert n into the i-th index of li
    li.insert(i, n)

    print(li)


if __name__ == "__main__":
    for n in range(11):
        li = [1, 2, 2, 3, 4, 6, 8, 9]
        print(li, n)
        insort(li, n)
        print()
```

```python
def calculate_grade(marks):
    if marks >= 80:
        grade = "A+"
    elif marks >= 70:
        grade = "A"
    elif marks >= 60:
        grade = "A-"
    elif marks >= 50:
        grade = "B"
    elif marks >= 40:
        grade = "C"
    else:
        grade = "F"

    return grade
```

```python
def calculate_grade_bisect(marks):
    grades = ["F", "C", "B", "A-", "A", "A+"]
    grade_thresholds = [40, 50, 60, 70, 80]

    i = bisect(grade_thresholds, marks)
    return grades[i]
```

```python
if __name__ == "__main__":
    for i in range(101):
        print(f"{i}: {calculate_grade(i)} (bisect: {calculate_grade_bisect(i)})")
        assert calculate_grade(i) == calculate_grade_bisect(i), i
```
