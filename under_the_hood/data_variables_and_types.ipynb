```python
print("Address of value 42: ", hex(id(42)))
x = 42
print("Address referred to by x: ", hex(id(x)))
print("Value of x: ", x)
print("-----")

y = x
print("Address referred to by x: ", hex(id(x)))
print("Address referred to by y: ", hex(id(y)))
print("-----")

x = 3.14
print("Address referred to by x: ", hex(id(x)))
print("Address referred to by y: ", hex(id(y)))
```

```python
print("Address of value 42: ", hex(id([42])))
x = [42]
print("Address referred to by x: ", hex(id(x)))
print("Value of x: ", x)
print("-----")

y = x
print("Address referred to by x: ", hex(id(x)))
print("Address referred to by y: ", hex(id(y)))
print("-----")

y = [3.14]
print("Address referred to by x: ", hex(id(x)))
print("Address referred to by y: ", hex(id(y)))
```
